```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder

onehot_cols = [
    "gender",
    "academic_work_impact"
]

ordinal_cols = [
    "stress_level"
]

onehot_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore" , sparse_output=False))
])

ordinal_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OrdinalEncoder(
        categories=[["low", "Medium", "High"]],
        handle_unknown="use_encoded_value",
        unknown_value=-1
        
    ))
])

preprocessor = ColumnTransformer([
    ("onehot", onehot_pipeline, onehot_cols),
    ("ordinal", ordinal_pipeline, ordinal_cols)
], remainder="drop")
```

#### Feature Enginnering

```python
import numpy as np

def create_features(df):
    df = df.copy()

    screen = df["daily_screen_time_hours"].clip(lower=1e-6)
    sleep = df["sleep_hours"].clip(lower=1e-6)
    opens = df["app_opens_per_day"].clip(lower=1e-6)

    # Combined usage
    df["leisure_screen_hours"] = (
        df["social_media_hours"] +
        df["gaming_hours"]
    )

    # Screen-time composition
    df["social_media_share"] = (
        df["social_media_hours"] / screen
    )

    df["gaming_share"] = (
        df["gaming_hours"] / screen
    )

    df["leisure_screen_share"] = (
        df["leisure_screen_hours"] / screen
    )

    # Weekend behaviour
    df["weekend_screen_change"] = (
        df["weekend_screen_time"] -
        df["daily_screen_time_hours"]
    )

    df["weekend_screen_ratio"] = (
        df["weekend_screen_time"] / screen
    )

    # Digital-life balance
    df["screen_sleep_ratio"] = (
        df["daily_screen_time_hours"] / sleep
    )

    df["screen_sleep_difference"] = (
        df["daily_screen_time_hours"] -
        df["sleep_hours"]
    )

    # Usage intensity
    df["notifications_per_screen_hour"] = (
        df["notifications_per_day"] / screen
    )

    df["app_opens_per_screen_hour"] = (
        df["app_opens_per_day"] / screen
    )

    df["notifications_per_app_open"] = (
        df["notifications_per_day"] / opens
    )

    return df.replace([np.inf, -np.inf], np.nan)
```

## Data Cleaning Complete Pipeline

```python
from sklearn import set_config
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import FunctionTransformer
from sklearn.impute import SimpleImputer

set_config(transform_output="pandas")

num_pipeline = Pipeline([
    ("imputer", SimpleImputer(
        strategy="mean",
        add_indicator=False
     ))
])
```

```python
encoding_pipeline = Pipeline([
    ("imputer", SimpleImputer(
        strategy="mean",
        add_indicator=False
     )),
    ("feature_engineering", FunctionTransformer(
        create_features,
        validate=False
    ))
])
```

```python
num_attribs = list(df_num)
cat_attribs = list(df_cat)

pre_processor = ColumnTransformer([
        ("num", num_pipeline, num_attribs),
        ("cat", preprocessor , cat_attribs )
    ])


pre_encoding = ColumnTransformer([
  ("encoding",encoding_pipeline, num_attribs),
      ("cat", preprocessor , cat_attribs )
    ])
```

```python
import lightgbm as lgb
from sklearn.pipeline import Pipeline
from sklearn.model_selection import cross_val_score

lightgbm_model = lgb.LGBMClassifier(
    objective="binary",
    n_estimators=1000,
    learning_rate=0.05,
    num_leaves=31,
    random_state=42,
    verbosity=-1
)

lightgbm_pipeline = Pipeline([
    ("preprocessing",  pre_encoding),
    ("model", lightgbm_model)
 ])
```

```python
param_distributions_lgb = {

    # # preprocessing
    # "preprocessing__num__imputer__strategy": [
    #     "mean",
    #     "median"
    # ],

    # LightGBM
    "model__n_estimators": [
        300, 500, 800, 1200
    ],

    "model__learning_rate": [
        0.02, 0.05, 0.1
    ],

    "model__max_depth": [
        4, 5, 6, 7, 8, 10
    ],

    "model__num_leaves": [
        15, 31, 63, 127
    ],

    "model__min_child_samples": [
        10, 20, 30, 50
    ],

    "model__reg_lambda": [
        0, 1, 5, 10, 20
    ]
}
```

```python
from sklearn.model_selection import StratifiedKFold, cross_val_score

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

scores = cross_val_score(
    lightgbm_pipeline,
    df,               
    removed_label,
    cv=cv,
    scoring="roc_auc",
    n_jobs=-1
)

print("Fold scores:", scores)
print("Mean ROC-AUC:", scores.mean())
print("Standard deviation:", scores.std())
```

```python
from sklearn.model_selection import RandomizedSearchCV

searchlgb = RandomizedSearchCV(
    estimator=lightgbm_pipeline,
    param_distributions=param_distributions_lgb,
    n_iter=15,
    scoring="roc_auc",
    cv=cv,
    random_state=42,
    n_jobs=1,
    verbose=1,
    refit=True
)

searchlgb.fit(df, removed_label)
```

```python
print("Best ROC-AUC:", searchlgb.best_score_)
print("Best parameters:", searchlgb.best_params_)
```
